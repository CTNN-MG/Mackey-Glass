```python
# --- Mackey-Glass Generator ---
def mackey_glass(t_max, tau=17, delta_t=1, beta=0.2, gamma=0.1, n=10,
                 x0=1.2, num_anomalies=20, window=20, scale=0.5, seed=42):
    """
    Mackey-Glass time series with anomalies.

    - num_anomalies: number of anomaly windows
    - window: size of each anomaly window
    - scale: multiplier for anomaly size
    - seed: random seed for reproducibility
    """
    t_len = int(t_max / delta_t)
    x = np.zeros(t_len)
    x[0] = x0
    for t in range(1, t_len):
        if t - tau >= 0:
            x[t] = x[t - 1] + delta_t * (beta * x[t - tau] / (1 + x[t - tau] ** n) - gamma * x[t - 1])
        else:
            x[t] = x[t - 1] - delta_t * gamma * x[t - 1]

    # --- Insert anomalies ---
    rng = np.random.default_rng(seed)
    anomaly_indices = rng.choice(range(window, t_len - window), num_anomalies, replace=False)
    for idx in anomaly_indices:
        noise = rng.uniform(-0.01, 0.01, size=window) * scale
        x[idx:idx+window] += noise

    return x
```
